```python
import pandas as pd
import numpy as np

data = pd.read_csv('C:/Users/Nitesh Kumhar/Desktop/ADST_nitesh/Case_Study_3/trends.csv')
display(data)

print(data.shape)
print(data.head())
print(data.info())
print(data.isnull().sum())
print(data.describe())
```

```python
q1 = data['rank'].quantile(0.25) 
print(f"Q1 : {q1}")

q3 = data['rank'].quantile(0.75) 
print(f"Q3 : {q3}")

iqr = q3 - q1
print(f"IQR : {iqr}")

lowerbound = q1 - 1.5 * iqr
upperbound = q3 + 1.5 * iqr

print(f"LowerLimit : {lowerbound}")
print(f"UpperLimit : {upperbound}")

idr_outliner = data[(data['rank'] < lowerbound) | (data['rank'] > upperbound )]
clear_outliner = data[(data['rank'] >= lowerbound) | (data['rank'] <= upperbound )]

print(f"Cleaned Dataframe rows: {len(clear_outliner)}")
print(idr_outliner)
print(idr_outliner['rank'])
```
