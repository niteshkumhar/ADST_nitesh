```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler 

data = {
    'Age' : [25,45,31,52,22],
}

df = pd.DataFrame(data)
print(df)


scaler = MinMaxScaler()
scaled = scaler.fit_transform(df[['Age']])
df_scaler = pd.DataFrame(scaled,columns=['Age_scaled'])

print(df_scaler)
```
